```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
import random

scaler = StandardScaler()

pd.set_option("display.max_rows", None)     # 행 제한 해제
pd.set_option("display.max_columns", None)  # 열 제한 해제
pd.set_option("display.width", None)        # 한 줄에 다 나오게
pd.set_option("display.max_colwidth", None) # 컬럼 길이 제한 해제

DATA_PATH = "./preprocess/03_processed_data/data_set.csv"
df = pd.read_csv(DATA_PATH, encoding="utf-8-sig")
```

```python
# 분석에 안 쓰는 컬럼 제외
drop_cols = ["년도", "월", "지역"]
df = df.drop(columns=[c for c in drop_cols if c in df.columns])

# 숫자형 변수만 선택
num_df_ori = df.select_dtypes(include=[np.number]).dropna(how="any")

num_df_ori = pd.DataFrame(
    scaler.fit_transform(num_df_ori),
    columns=num_df_ori.columns,
    index=num_df_ori.index
)

# 피어슨 상관계수 행렬
corr = num_df_ori.corr(method="pearson")
```

```python
def highlight_high_values(val):
    if abs(val) >= 0.8:
        return 'background-color: yellow; color: black;'
    return ''

# 적용
corr.style.applymap(highlight_high_values)
```

```python
# 행렬을 긴 포맷으로 변환
corr_long = (
    corr.reset_index()
        .melt(id_vars="index", var_name="변수2", value_name="상관계수")
        .rename(columns={"index": "변수1"})
)
print(len(corr_long))
# 자기 자신과의 상관계수 제거
corr_long = corr_long[corr_long["변수1"] != corr_long["변수2"]]

# 변수1, 변수2 쌍 정렬해서 중복 제거
corr_long[["변수1", "변수2"]] = np.sort(corr_long[["변수1", "변수2"]], axis=1)
corr_long = corr_long.drop_duplicates(subset=["변수1", "변수2"])

# 상관계수 기준 정렬
corr_long = (
    corr_long[corr_long["상관계수"].abs() >= 0.8]  
    .sort_values(by=["변수1", "상관계수"], ascending=[True, False])  # 변수1 → 상관계수 순으로 정렬
    .reset_index(drop=True)
)

corr_long
```

```python
# 3) 종속변수 / 독립변수 분리
y = num_df_ori["전기차 등록수"]   
X = num_df_ori.drop(columns=["전기차 등록수"])     


#VIF 계산
X_const = add_constant(X)
vif_df = pd.DataFrame()
vif_df["변수"] = X_const.columns
vif_df["VIF"] = [variance_inflation_factor(X_const.values, i) 
                 for i in range(X_const.shape[1])]

#상수항 제거
vif_df = vif_df[vif_df["변수"] != "const"].reset_index(drop=True)
vif_df
```

```python


num_df = num_df_ori.copy()

# 파생변수 생성

# 월별 평균 기온의 차이
num_df["월 기온차이"] = num_df["평균최저기온(°C)"] - num_df["평균최고기온(°C)"]

# 연료별 차량 증가 비율 
num_df["수소차 증가 비율"] = num_df["수소차_순증가량"] / num_df["차량전체_순증가량"]
num_df["하이브리드차 증가 비율"] = num_df["하이브리드차_순증"] / num_df["차량전체_순증가량"]
num_df["휘발유차 증가 비율"] = num_df["휘발유차_순증가량"] / num_df["차량전체_순증가량"]
num_df["경유차 증가 비율"] = num_df["경유차_순증가량"] / num_df["차량전체_순증가량"]
num_df["전기차 증가 비율"] = num_df["전기차 순증가량"] / num_df["차량전체_순증가량"]

# 충전기 종류 비율
num_df["충전기 종류 비율"] = num_df["완속충전기_누적대수"] / num_df["급속충전기_누적대수"] 
num_df["신규설치 충전기 종류 비율 "] = num_df["완속충전기_신규설치"] / num_df["급속충전기_신규설치"]

# 연료 가격 평균
num_df["휘발유 디젤 가격 평균"] = (num_df["휘발유가격"] + num_df["디젤가격"]) / 2

# 다중공선성 심한 변수 제거
num_df.drop(columns=[
    "완속충전기_신규설치", 
    "급속충전기_신규설치", 
    "전체충전기_신규설치", 
    "완속충전기_누적대수", 
    "급속충전기_누적대수",
    "평균최저기온(°C)",
    "평균최고기온(°C)",
    "차량전체_등록수",
    "1회충전주행거리",
    "하이브리드차_등록수",
    "수소차_등록수",
    "출고대수",
    "휘발유차_등록수",
    "경유차_등록수",
    "수소차_순증가량",
    "하이브리드차_순증",
    "휘발유차_순증가량",
    "경유차_순증가량",
    "차량전체_순증가량",
    "미세먼지10 수치",
    "휘발유가격",
    "디젤가격",
    "cpi",
    "전기차 순증가량"
    ,
    "휘발유차 증가 비율",
    "경유차 증가 비율"
    ], 
    inplace=True)


# 피어슨 상관계수 행렬
corr = num_df.corr(method="pearson")

corr.style.applymap(highlight_high_values)
```
